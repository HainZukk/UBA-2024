Para crear un diccionario dentro de un diccionario es necesario saber la sintaxis
Clave:valor

```python
diccionario = {
    "user":"Tobias",
    "password" : 1234
}
```

Para hacer un diccionario dentro de un diccionario es cuestion de hacer lo siguiente

```python
second_diccionario = {
    "user" : "Tobias",
    "password" : 1234,
    "info_plus" : {  #Debemos agregar un valor y agregar {} en el mismo formato key : valor
        "city" : "Grand Bourg",
        "direction" : "Nicolas Mascardi"
    }
    
}
```

Con los diccionarios podemos hacer infinidades de cosas como iterar sobre ellos o guardar informacion en datos csv o txt etc.

```python
second_diccionario.keys() # nos permite ver todas las keys dentro del diccionario 
```

```python
second_diccionario.values() # nos permite ver todos los valores asignados en el diccionario
```
